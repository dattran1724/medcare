```python
import pyreadstat
import pandas as pd

file_path = r"D:\2026\MedCare\ML\Dataset\Raw\h16f2c.sav"

# Chỉ đọc metadata
_, meta = pyreadstat.read_sav(
    file_path,
    metadataonly=True
)

# Tạo bảng metadata
metadata_df = pd.DataFrame({
    "variable": meta.column_names,
    "label": meta.column_labels
})

# Lưu CSV
output_path = r"D:\2026\MedCare\ML\Dataset\h16f2c_metadata.txt"

metadata_df.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig"
)

print("Đã lưu:", output_path)
print("Số biến:", len(metadata_df))
```
